- Why should a test set not be used during training?​
A test set should not be used during training in order to avoid leakage that can lead to overfitting

- What is data leakage? Give one example.​
Data leakage is when same data in validation or test is used during the final model training

- Why can accuracy be misleading?​
Accuracy can be misleading because it only shows the overall number of correct predictions.

- What is the difference between a feature and a label?​
The difference between a feature and a label is that a feature is the data we give the machine learning model to help it make a prediction, while the label is the correct answer that the model is trying to predict. feature is the input while the label is the output.

- What does overfitting look like?​
Overfitting is when the model memorises the training data instead of learning the general pattern so model performs very well on the training data but doesn't perform so well on the validation or test data

- Why should we build a simple baseline first?​
A baseline is like a starting point. It helps us measure whether our machine learning model is improving or not.

- What is the difference between loss and metric?​
loss - we use it during training. It is a mathematical signal that tells the model how wrong it is during training. It is for the model.
metric - It is used to judge performance. It is for the human.

- What should an error analysis include?​
The type of errors the model is making, Why the errors are happening, Which errors happen most often, Possible improvements
